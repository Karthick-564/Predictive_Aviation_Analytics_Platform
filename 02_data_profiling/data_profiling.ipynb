# Missing Value Analysis

```python
missing_counts = df.isnull().sum()
missing_df = pd.DataFrame({
    "Column_Name": missing_counts.index,
    "Missing_Count": missing_counts.values
}).sort_values(by="Missing_Count", ascending=False)

missing_df
```

#  Missing Value Percentage

```python
missing_df["Missing_Percentage"] = (missing_df["Missing_Count"] / len(df)) * 100

missing_df
```

#  Duplicate Record Check

```python
duplicate_count = df.duplicated().sum()
duplicate_count
```

### Duplicate Record Observation
- Duplicate rows indicate redundancy in real-world flight data.
- These will be handled during the data cleaning phase (Week 2).


# Save Data Quality Report

```python
missing_df.to_csv("../Reports/Data_Quality_Report.csv", index=False)
print(" Data Quality Report saved successfully.")
```

```python
data_quality_summary = {
    "Total Rows": df.shape[0],
    "Total Columns": df.shape[1],
    "Total Missing Values": df.isnull().sum().sum(),
    "Duplicate Rows": duplicate_count
}
data_quality_summary
```

```python
# Save summary
pd.DataFrame(
    list(data_quality_summary.items()),
    columns=["Metric", "Value"]
).to_csv("../Reports/Data_Quality_Summary.csv", index=False)
```
